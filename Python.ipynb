````python
### 📘 Python Mastery Notebook: Architecture, Data Structures, Functions & OOP

This notebook synthesizes core Python concepts, covering architecture, environments, data structures, control flow, functional programming, and object-oriented programming [1, 2].

---

### **Section 1: Python Architecture and Execution Model**

#### **1.1 Python Architecture Overview**

Python utilizes a **hybrid execution approach** [3-5]. Source Code (`.py` file) is compiled into **Bytecode** (`.pyc` files), which is then executed line by line by the **Python Virtual Machine (PVM)** [3, 4, 6]. The PVM manages objects using a Memory Manager and Garbage Collector [3, 6]. Most core Python functions are implemented in C for flexibility [6].

| Component | Role | Source |
| :--- | :--- | :--- |
| Compiler | Converts source code to bytecode [6]. | [4, 6] |
| Bytecode | Low-level instructions (not machine code), stored in `.pyc` files [3, 5]. | [3, 5] |
| PVM | Executes bytecode line by line [4, 6]. | [4, 6] |
| Memory Manager | Allocates, manages, and cleans memory (Object Heap) [6, 7]. | [6, 7] |

#### **1.2 Hybrid Execution Flow**

The flow is: Compiled to Bytecode $\rightarrow$ Interpreted by PVM [5]. If Python were purely interpreted, it would read and execute code line by line without a separate compilation step [8].

```python
# Execution flow: Source Code -> Bytecode -> PVM executes [3, 5]

# Example: Inspecting Bytecode using the dis module [9]
import dis
# Bytecode instructions: LOAD_NAME, LOAD_CONST, CALL_FUNCTION, RETURN_VALUE [9]
dis.dis("print('Hello, Aravind!')")

--------------------------------------------------------------------------------
Section 2: Environment Management (venv and Conda)
Environments isolate project dependencies to prevent "dependency hell".
2.1 venv (Python's Built-in Virtual Environment)
venv uses the system Python interpreter and is ideal for lightweight projects like web development or scripts.
# Creating, activating, and managing a venv environment [13]
print("# Create a new env")
print("python -m venv myenv")

# Activating (Windows PowerShell) [13]
print("# Activate")
print("myenv\\Scripts\\activate")

# Deactivating [13]
print("# Deactivate")
print("deactivate")

# Saving and reinstalling dependencies [13]
# pip install requests
# pip freeze > requirements.txt
# pip install -r requirements.txt
2.2 Conda Environments (Anaconda/Miniconda)
Conda acts as both a package manager and an environment manager. It can install Python itself and non-Python dependencies (C, CUDA, R). It uses a SAT solver for advanced dependency resolution, making it perfect for Data Science and AI/ML.
# Creating a Conda environment [14]
print("# Create new environment with Python 3.11")
print("conda create --name ai-lab python=3.11")

# Activating and Deactivating [14]
print("# Activate")
print("conda activate ai-lab")
print("# Deactivate")
print("conda deactivate")

# Installing packages (can install Python and system libs) [14, 15]
print("conda install numpy pandas matplotlib")

# Exporting and Recreating the environment [14]
print("# Export to YAML file")
print("conda env export > environment.yml")
print("# Recreate environment")
print("conda env create -f environment.yml")
2.3 Co-existence Demo
Both venv and Conda environments can exist without conflict on the same machine because each is an isolated folder containing its own dependencies and interpreter.
# Demo showing separate Django versions co-existing [18, 19]
print("# Venv (Web Project): Django 3.2 installed")
# python -m venv webenv
# webenv\Scripts\activate
# pip install django==3.2
# python -m django --version # Output: 3.2

print("# Conda (ML Project): Django 4.0 installed")
# conda activate ml-env
# conda install django=4.0
# python -m django --version # Output: 4.0

--------------------------------------------------------------------------------
Section 3: Variables, Data Types & Memory Management
3.1 Basic Data Types and Type Casting
Python is dynamically typed. Basic types include int, float, str, and bool. Type casting uses built-in functions like int(), float(), str(), and bool().
# Type Casting Examples [22]
age_str = "25"
age_num = int(age_str)
print(age_num + 5)      # 30

pi = 3.99
print(int(pi))          # 3 (truncates decimals)

# Boolean casting: 0, empty strings (""), [], and None are Falsy [22, 23]
print(bool(0))    # False
print(bool("Hi")) # True
3.2 Type Hints (Function Annotations)
Type hints improve readability and enable static analysis by tools like mypy. They are not enforced at runtime.
from typing import List, Dict, Optional, Union, Any, Callable, Tuple, Set
# For Python 3.9+, standard list[int] syntax is preferred [28]

def add_numbers(a: int, b: int) -> int:
    return a + b

def user_profile() -> Dict[str, Union[str, int]]:
    # Union allows multiple types [29]
    return {"name": "Aravind", "age": 25}

def get_discount(price: float, coupon: Optional[str]) -> float:
    # Optional[T] is shorthand for Union[T, None] [29]
    return price * 0.9 if coupon == "NEW10" else price

def apply_operation(x: int, y: int, func: Callable[[int, int], int]) -> int:
    # Callable hint: input types (int, int), return type (int) [29]
    return func(x, y)

point: Tuple[int, int] = (10, 20)
3.3 Memory Management: Reference Counting & GC
Python primarily uses reference counting; memory is freed when the reference count drops to zero. The Garbage Collector (GC) handles cyclic references (objects pointing to each other).
import sys, gc

# Example 1: Reference Counting [30]
x = [1, 3, 6]
y = x
# sys.getrefcount(x) increases as more names point to it.

del y
# Ref count decreases. When it hits zero, the object is destroyed.

# Example 2: Cyclic References and GC [32, 33]
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

n1 = Node("A")
n2 = Node("B")
n1.next = n2
n2.next = n1  # Cycle created
del n1, n2 # Objects are not freed because ref count > 0 due to the cycle [33]
unreachable = gc.collect()
print(f"Unreachable objects collected: {unreachable}")

# del statement removes the name/reference, not the object itself [34]
a = 100
b = a
del a # Name 'a' is removed
print(f"b still exists: {b}")
3.4 Context Managers and Weak References
Context Managers (via with statement) guarantee safe resource management (setup via __enter__, cleanup via __exit__), regardless of exceptions.
Weak References (weakref) allow referencing an object without increasing its reference count, ensuring the object can still be garbage collected.
import weakref
from contextlib import contextmanager

# 1. Standard Context Manager (File) [37]
with open("temp_data.txt", "w") as f:
    f.write("Test log") # File is automatically closed by __exit__() [37]

# 2. Custom Function-based Context Manager (using yield) [39, 41]
@contextmanager
def managed_resource():
    print("Setup (Before yield)")
    yield "Resource Handle" # Control passes to 'with' block [41]
    print("Teardown (After yield)")

with managed_resource() as r:
    print("Using resource:", r)

# 3. Weak Reference Example [40]
class Data: pass
obj = Data()
r = weakref.ref(obj) # Weak reference does not increase refcount
del obj
print(f"Object accessible after del: {r()}") # None (object destroyed)
3.5 Shallow vs Deep Copying
Operation
Outer Object
Inner Objects
Source
Assignment =
Shared
Shared
Shallow Copy
New
Shared
Deep Copy
New
New
Shallow copy (copy.copy()) creates a new container but shares references to inner, nested objects. Deep copy (copy.deepcopy()) recursively duplicates all nested objects, ensuring full independence.
import copy

# Nested structure (list containing mutable lists) [44]
a = [[1, 3], [6, 46]]

# 1. Shallow Copy (Inner list SHARED) [44]
b_shallow = copy.copy(a)
b_shallow = 99
print(f"A (Shallow Copy modification): {a}") # [[3, 34], [6, 46]]

# Reset A
a = [[1, 3], [6, 46]]

# 2. Deep Copy (Inner list COPIED) [45]
b_deep = copy.deepcopy(a)
b_deep = 999
print(f"A (Deep Copy modification): {a}") # [[1, 3], [6, 46]]

--------------------------------------------------------------------------------
Section 4: Operators and Control Flow
4.1 Arithmetic and Comparison Operators
Python differences from Java/C++: True division (/) always returns a float, and Floor division (//) provides the integer quotient. Exponentiation uses **. Comparisons can be chained (e.g., 5<x<10).
# Arithmetic Operators [47]
print(f"True Division: 5 / 2 = {5 / 2}") # 2.5
print(f"Floor Division: 5 // 2 = {5 // 2}") # 2
print(f"Exponentiation: 2 ** 3 = {2 ** 3}") # 8

# Chained Comparisons [49]
x = 7
print(f"5 < x < 10: {5 < x < 10}")
4.2 Logical and Identity Operators
Python uses keywords (and, or, not). Identity (is) checks if two variables refer to the exact same object in memory, unlike == which checks value equality.
# Logical Operators [51]
x = 5
print(f"x > 0 and x < 10: {x > 0 and x < 10}")

# Identity vs Equality [52]
a = [1, 3]
b = [1, 3]
c = a
print(f"a == b (Value Equality): {a == b}") # True
print(f"a is b (Identity Check): {a is b}") # False (different objects)
print(f"a is c: {a is c}") # True (same object)

# Membership Operator ('in') [53]
nums = [1, 3]
print(f"2 in nums: {2 in nums}")
4.3 The Walrus Operator (:=)
The assignment expression (:=) allows assigning a value to a variable within an expression.
# Walrus in a conditional statement [54]
if (n := len("Python")) > 5:
    print(f"String is too long, length: {n}")

# Walrus in a list comprehension [55]
nums = [56-60]
result = [half for n in nums if (half := n // 2) > 10]
# Walkthrough: 10/2=5(X), 25/2=12(V), 30/2=15(V), 5/2=2(X), 60/2=30(V) [61]
print(f"Walrus comprehension result: {result}")
# Output: [58, 62, 63] [64]
4.4 Control Flow: if/elif/else and Loops
Python uses indentation to define code blocks, and elif instead of else if. for loops iterate over sequences (like Java's enhanced for-each).
# if/elif/else syntax [65]
x = 10
if x > 0:
    print("Positive")
elif x == 0:
    print("Zero")
else:
    print("Negative")

# Iterating using range() [66]
print("Range loop (2, 10, step 2):")
for i in range(2, 10, 2):  # 2, 4, 6, 8 (stop index 10 is exclusive) [66]
    print(i)
4.5 Unique Python Feature: else with Loops
The else clause attached to a for or while loop executes only if the loop finishes normally (i.e., not terminated by a break statement). This is often used for "not found" checks.
# Example 1: Loop finishes normally (else runs) [69]
for i in range(3):
    print(i)
else:
    print("Loop finished without break") # This runs

# Example 2: Loop exits via break (else is skipped) [69]
for i in range(5):
    if i == 3:
        break
    print(i)
else:
    print("Loop finished without break") # This is skipped

--------------------------------------------------------------------------------
Section 5: Comprehensions and Generators
5.1 List Comprehensions
List comprehensions provide a concise way to create lists: [expression for item in iterable if condition].
nums = [1, 3, 6, 46, 59]

# Example 1: Squares of numbers [71]
squares = [n*n for n in nums]
print(f"Squares: {squares}")

# Example 2: Filtering evens [71]
evens = [n for n in nums if n % 2 == 0]
print(f"Evens: {evens}")

# Example 3: Nested Loops (Flattening) [72, 73]
matrix = [[1, 3, 6], [46, 59, 74]]
flat = [num for row in matrix for num in row]
print(f"Flattened list: {flat}")
5.2 Dictionary and Set Comprehensions
Dictionary: {key_expr: value_expr for item in iterable}. Set: {expr for item in iterable}.
# Dictionary Comprehension: Key-Value Pairs [75]
nums = [1, 3, 6]
squares_dict = {n: n*n for n in nums}
print(f"Dict Squares: {squares_dict}")

# Set Comprehension: Unique items [77]
nums_dupes = [1, 3, 3, 6, 6, 46]
unique_squares = {n*n for n in nums_dupes}
print(f"Set Squares (unordered): {unique_squares}")
5.3 Generator Expressions and Memory Efficiency
Generator expressions use () instead of []. They use lazy evaluation, producing items one at a time, making them highly memory-efficient for large datasets or streams.
import sys

# List Comprehension (Eager - stores all items) [80]
list_squares = [n*n for n in range(5)]

# Generator Expression (Lazy - computes on demand) [78, 80]
gen_squares = (n*n for n in range(5))

print(f"List size (approx): {sys.getsizeof(list_squares)} bytes")
print(f"Generator size (tiny recipe): {sys.getsizeof(gen_squares)} bytes")

# Generator pipeline for memory efficiency (only one item in memory at a time) [81]
nums = range(1_000_000)
# Sum of cubes of numbers divisible by 7 (lazy computation)
total = sum(n**3 for n in nums if n % 7 == 0)

--------------------------------------------------------------------------------
Section 6: Data Structures: List, Tuple, Set, Dictionary
6.1 Lists (Mutable Sequences)
Lists are ordered, mutable collections. Operations like inserting/deleting at the front (insert(0, x), pop(0)) are O(n) due to shifting elements.
# Slicing and Negative Indexing [85]
fruits = ["apple", "banana", "cherry", "date"]
print(f"Last element: {fruits[-1]}")
print(f"Reversed list: {fruits[::-1]}")

# Adding and Removing [86, 87]
lst = [1, 3]
lst.append(3)           # [1, 3, 6]
lst.insert(1, 5)        # [1, 3, 6, 59]
lst.extend([8, 74])      # [1, 3, 6, 8, 59, 74]
lst.remove(5)           # Removes first 5 [87]
lst.pop()               # Removes last (7) [88]
print(f"List after ops: {lst}")

# Sorting: sort() mutates in place; sorted() returns a new list [89]
nums = [1, 6, 46]
nums.sort()
print(f"Sorted in place: {nums}")
6.2 Tuples (Immutable Sequences)
Tuples are ordered, immutable sequences. They are faster and use less memory than lists, and are hashable (can be dictionary keys).
# Creating Tuples (Note: comma required for single element: (42,) ) [92, 93]
t1 = (1, 2, 3)
t_packed = 4, 5, 6 # Tuple packing [92]

# Immutability check
# t1 = 10 # Raises TypeError [94]

# Tuple Unpacking [94]
x, y, z = t1
print(f"Unpacked: {x}, {y}, {z}")

# Extended Unpacking (using *) [91]
data = (1, 2, 3, 4, 5)
a, *b, c = data
print(f"Extended Unpacking: a={a}, b={b} (list), c={c}")

# Swapping variables without temp [95]
a, b = 10, 20
a, b = b, a
print(f"Swapped: a={a}, b={b}")
6.3 Sets and Frozen Sets
Sets are unordered collections of unique elements, supporting fast membership checks (O(1)). Sets are mutable. Frozen Sets are immutable and hashable, usable as dictionary keys.
# Creating Sets [96]
fruits = {"apple", "banana", "cherry", "apple"}
print(f"Set (duplicates removed): {fruits}")

# Set Operations (Mathematical) [100, 101]
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(f"Union (a | b): {a | b}")         # All unique elements [100]
print(f"Intersection (a & b): {a & b}") # Common elements [100]
print(f"Difference (a - b): {a - b}")   # In a but not in b [102]
print(f"Subset (a <= b): {a <= b}")     # Check if a is contained in b [103]

# Frozen Set Example [104]
fs = frozenset([1, 3])
# fs.add(3) # Raises AttributeError [99]
d = {fs: "data"} # Can be used as a key [99]
6.4 Dictionaries (Key-Value Pairs)
Dictionaries are mutable collections of key-value pairs. Keys must be unique and immutable. Dictionaries maintain insertion order (Python 3.7+).
# Creating and Accessing [108]
student = {"name": "Alice", "age": 22, "major": "CS"}
print(f"Access (direct): {student['name']}")
print(f"Access (safe): {student.get('grade', 'N/A')}") # Use .get() to avoid KeyError [109]

# Updating and Adding [109]
student["age"] = 23 # Update
student["email"] = "alice@example.com" # Add new
student.update({"major": "Physics"}) # Merge/overwrite [109]

# Deletion [110]
age = student.pop("age") # Remove and return value (safer than del)
del student["major"] # Delete specific key
print(f"Student keys left: {student.keys()}")

# Dictionary Comprehension [111]
nums = [1, 3, 6]
squares_comp = {n: n**2 for n in nums}
print(f"Dict Comp: {squares_comp}")
6.5 Advanced Dictionary Tools (collections)
These tools provide enhanced functionality over standard dicts.
from collections import defaultdict, Counter, ChainMap, namedtuple
from types import MappingProxyType
from dataclasses import dataclass, field

# 1. defaultdict: Provides default value for missing keys [113, 114]
count = defaultdict(int)
words = ["apple", "banana", "apple"]
for w in words: count[w] += 1
print(f"defaultdict: {count}")

# 2. Counter: Subclass for counting hashable objects [115, 116]
freq = Counter(words)
print(f"Counter (Most common): {freq.most_common(1)}")

# 3. ChainMap: Combines multiple dictionaries (searches layers in order) [117, 118]
defaults = {"theme": "light"}
user_settings = {"theme": "dark"}
config = ChainMap(user_settings, defaults)
print(f"ChainMap theme: {config['theme']}")

# 4. Named Tuples: Immutable record types [119, 120]
Point = namedtuple('Point', ['x', 'y'])
p = Point(10, 20)
print(f"Named Tuple: {p.x}")

# 5. Dataclasses: Structured data with auto-generated methods [121, 122]
@dataclass
class Person:
    name: str
    age: int = 18
    skills: list = field(default_factory=list) # Use default_factory for mutable defaults [123]

--------------------------------------------------------------------------------
Section 7: Functions and Advanced Concepts
7.1 Function Definition and Arguments
Functions are declared with def. They are first-class citizens (can be assigned, passed, returned). Python supports multiple returns (packed as a tuple).
# Multiple Returns (Tuple Packing/Unpacking) [130]
def min_max(numbers):
    return min(numbers), max(numbers) # Returns a tuple
low, high = min_max([1, 3, 9])
print(f"Min: {low}, Max: {high}")

# Default Parameters and Keyword Arguments [127, 130]
def describe_pet(name, species="Dog"): # Default
    print(f"{name} is a {species}")

describe_pet(species="Cat", name="Whiskers") # Keyword Arguments (order irrelevant) [131]
describe_pet("Buddy") # Uses default
7.2 Variable-Length Arguments (*args, **kwargs)
*args collects extra positional arguments into a tuple. **kwargs collects extra keyword arguments into a dictionary.
# *args: collects positional arguments into a tuple [132]
def add_all(*numbers):
    return sum(numbers)
print(f"Sum: {add_all(1, 2, 3, 4)}")

# **kwargs: collects keyword arguments into a dict [132]
def show_profile(name, **info):
    print(f"Name: {name}. Other info: {info}")
show_profile(name="Aravind", age=25, role="Dev")
7.3 The Mutable Default Parameter Trap ⚠️
Mutable objects (lists, dicts, sets) should not be used as default parameters, as they are shared across all function calls.
# ❌ INCORRECT: Mutable default [] is shared [135]
# def add_item_bad(item, basket=[]):
#     basket.append(item)
# print(add_item_bad("apple")) # ['apple']
# print(add_item_bad("banana")) # ['apple', 'banana'] <-- shared list! [135]

# ✅ CORRECT: Use None and initialize inside the function [136]
def add_item_safe(item, basket=None):
    if basket is None:
        basket = [] # New list created on each call
    basket.append(item)
    return basket

print(f"Safe call 1: {add_item_safe('apple')}")
print(f"Safe call 2: {add_item_safe('banana')}")
7.4 Lambda Functions and Functional Programming
Lambda functions are small, anonymous, one-line functions (lambda args: expression). They are often used with Higher-Order Functions (HOFs) like map, filter, and sorted.
from functools import reduce

# Lambda syntax: lambda arguments: expression [137]
square = lambda x: x**2
print(f"Lambda square(5): {square(5)}")

nums = [1, 3, 6, 46]
# map: applies lambda to all elements [138]
squared = list(map(lambda x: x**2, nums))

# filter: keeps elements returning True [138]
evens = list(filter(lambda x: x % 2 == 0, nums))

# sorted: uses lambda as a key function to sort by custom criteria [140]
points = [(1, 5), (2, 3), (4, 1)]
sorted_points = sorted(points, key=lambda p: p[1]) # Sort by y-value
print(f"Sorted by key: {sorted_points}")
7.5 Higher-Order Functions (HOFs) and Closures
A HOF takes or returns a function. Closures are inner functions that remember variables from their outer (enclosing) scope even after the outer function finishes executing. This enables state maintenance and decorators.
# HOF: Function that returns a function (Closure Example) [146]
def make_multiplier(n): # n is captured by the closure
    def multiplier(x):
        return x * n # Inner function remembers 'n'
    return multiplier

double = make_multiplier(2)
print(f"Double 10: {double(10)}") # 20

# Closure for maintaining state (Counter Example) [147]
def counter():
    count = 0
    def increment():
        nonlocal count # Needed to modify enclosing variable [147]
        count += 1
        return count
    return increment

c = counter()
print(f"Counter: {c()}")
print(f"Counter: {c()}") # State persists [148]
7.6 Decorators and Decorator Factories
Decorators are HOFs that wrap functions to add functionality using the @ syntax. Decorators with arguments (like @repeat(3)) are called Decorator Factories.
import time
from functools import wraps

# Example 1: Basic Decorator (Wrapping behavior) [152]
def greet_decorator(func):
    @wraps(func) # Preserves name/docstring [153]
    def wrapper(*args, **kwargs):
        print("Before the function runs")
        result = func(*args, **kwargs)
        print("After the function runs")
        return result
    return wrapper

@greet_decorator
def say_hello():
    print("Hello!")
say_hello()

# Example 2: Decorator Factory (Decorator with Arguments) [154]
def repeat(n):
    def decorator(func): # Closure remembers 'n' [154]
        @wraps(func)
        def wrapper(*args, **kwargs): # Handles function arguments
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator

@repeat(3)
def greet():
    print("Hi")
greet() # Runs 'Hi' three times [151]

--------------------------------------------------------------------------------
Section 8: Modules, Packages, and pip
8.1 Modules and Importing
A module is a single .py file. A package is a directory containing modules, marked by an __init__.py file.
import math
from datetime import datetime, timedelta

print(f"Square root of 16: {math.sqrt(16)}") # Basic import [157]

now = datetime.now()
future = now + timedelta(days=7)
print(f"One week later: {future}") # Standard Library usage [159]
8.2 Writing Custom Modules and Packages
The __init__.py file is essential for marking a folder as a package and is often used to expose a clean public API via re-exporting functions (e.g., from .algebra import add).
# Example: Using the __name__ guard [161]
# Runs only if the file is executed directly (python file.py) [161]
# if __name__ == "__main__":
#     print("Testing module...")

# Example: Re-exporting in __init__.py [160]
# In mymath/__init__.py:
# from .algebra import add
# from .geometry import area_circle
# Usage: from mymath import add, area_circle
8.3 Package Management with pip
pip is Python's package installer, fetching libraries from PyPI. Virtual environments (venv or Conda) are essential for isolating dependencies.
# Installing and managing packages [162]
print("pip install requests")
print("pip install requests==2.31.0")

# Reproducibility (locking dependency versions) [165]
print("# Freeze installed packages to a file")
print("pip freeze > requirements.txt")
print("# Install from the file")
print("pip install -r requirements.txt")

--------------------------------------------------------------------------------
Section 9: Object-Oriented Programming (OOP)
9.1 Classes, Attributes, and the Constructor
The constructor is defined by __init__(self, ...). self refers to the current instance (like this in Java). Class variables are shared across instances.
class Dog:
    species = "Canine" # Class variable (shared) [169]

    def __init__(self, name, breed):
        self.name = name   # Instance attribute [166]
        self.breed = breed

    def bark(self):
        return f"{self.name} says Woof!"

dog1 = Dog("Buddy", "Labrador")

# Modification via Class Name affects all instances [170]
Dog.species = "Domestic Canine"
print(f"{dog1.name} is a {dog1.species}")
9.2 Inheritance, Polymorphism, and Encapsulation
Python supports both single and multiple inheritance. Encapsulation uses naming conventions: Protected (_attribute) and Private (__attribute, which is name-mangled). Abstraction uses Abstract Base Classes (ABC) and @abstractmethod to enforce interfaces.
from abc import ABC, abstractmethod

# Abstract Base Class (Abstraction) [176, 177]
class MLModel(ABC):
    def __init__(self, name):
        self._trained = False # Protected attribute (convention) [172]

    @abstractmethod # Forces subclasses to implement [175]
    def train(self, data):
        pass

# Utility Class (for Multiple Inheritance) [177-179]
class Logger:
    def log(self, message):
        print(f"[LOG]: {message}")

class LinearRegressionModel(MLModel, Logger): # Multiple Inheritance [171, 180]
    def __init__(self, name, learning_rate):
        super().__init__(name) # Calls parent (MLModel) constructor [181, 182]
        self.__weights = None # Private attribute (name-mangled) [173]

    def train(self, data):
        self._trained = True
        self.log(f"{self.name} trained.") # Use inherited method

# Polymorphic Function [183]
def evaluate_model(model: MLModel, data, input_data):
    # Works regardless of subclass type
    model.train(data)
    # Polymorphism: calls the appropriate train() method
9.3 Advanced OOP Features
a) Magic Methods (Dunder Methods)
Methods like __len__ and __repr__ allow classes to emulate built-in types.
class Dataset:
    def __init__(self, name, samples):
        self._samples = samples

    def __len__(self): # Defines behavior for len(obj) [185]
        return len(self._samples)

    def __repr__(self): # Unambiguous string representation for debugging [185]
        return f"Dataset(name='{self.name}', samples={len(self)})"

ds = Dataset("MyData", ["a", "b"])
print(f"Length: {len(ds)}") # Calls __len__
b) Class Method vs Static Method
Decorator
Argument
Purpose
Source
@classmethod
cls
Alternate constructors, uses class state
@staticmethod
None
Utility functions, no access to instance or class
class DataFactory:
    @staticmethod
    def validate_sample(sample): # Utility function [188]
        return isinstance(sample, str)

    @classmethod
    def from_config(cls, config_dict): # Alternate constructor [187]
        return cls(config_dict["name"], config_dict["data"])
c) Property Decorators (@property, @setter)
Provide controlled access (getters and setters) to internal attributes, making methods behave like attributes. You can use @property alone for read-only access.
class Circle:
    def __init__(self, radius):
        self._radius = radius # Internal storage

    @property
    def radius(self): # Acts as the getter [189]
        return self._radius

    @radius.setter
    def radius(self, value): # Controlled setter (validation logic) [190]
        if value < 0:
            raise ValueError("Radius cannot be negative")
        self._radius = value

c = Circle(5)
print(f"Radius (property access): {c.radius}")
c.radius = 10 # Uses the setter method### 📘 Python Mastery Notebook: Architecture, Data Structures, Functions & OOP

This notebook synthesizes core Python concepts, covering architecture, environments, data structures, control flow, functional programming, and object-oriented programming [1, 2].

---

### **Section 1: Python Architecture and Execution Model**

#### **1.1 Python Architecture Overview**

Python utilizes a **hybrid execution approach** [3-5]. Source Code (`.py` file) is compiled into **Bytecode** (`.pyc` files), which is then executed line by line by the **Python Virtual Machine (PVM)** [3, 4, 6]. The PVM manages objects using a Memory Manager and Garbage Collector [3, 6]. Most core Python functions are implemented in C for flexibility [6].

| Component | Role | Source |
| :--- | :--- | :--- |
| Compiler | Converts source code to bytecode [6]. | [4, 6] |
| Bytecode | Low-level instructions (not machine code), stored in `.pyc` files [3, 5]. | [3, 5] |
| PVM | Executes bytecode line by line [4, 6]. | [4, 6] |
| Memory Manager | Allocates, manages, and cleans memory (Object Heap) [6, 7]. | [6, 7] |

#### **1.2 Hybrid Execution Flow**

The flow is: Compiled to Bytecode $\rightarrow$ Interpreted by PVM [5]. If Python were purely interpreted, it would read and execute code line by line without a separate compilation step [8].

```python
# Execution flow: Source Code -> Bytecode -> PVM executes [3, 5]

# Example: Inspecting Bytecode using the dis module [9]
import dis
# Bytecode instructions: LOAD_NAME, LOAD_CONST, CALL_FUNCTION, RETURN_VALUE [9]
dis.dis("print('Hello, Aravind!')")

--------------------------------------------------------------------------------
Section 2: Environment Management (venv and Conda)
Environments isolate project dependencies to prevent "dependency hell".
2.1 venv (Python's Built-in Virtual Environment)
venv uses the system Python interpreter and is ideal for lightweight projects like web development or scripts.
# Creating, activating, and managing a venv environment [13]
print("# Create a new env")
print("python -m venv myenv")

# Activating (Windows PowerShell) [13]
print("# Activate")
print("myenv\\Scripts\\activate")

# Deactivating [13]
print("# Deactivate")
print("deactivate")

# Saving and reinstalling dependencies [13]
# pip install requests
# pip freeze > requirements.txt
# pip install -r requirements.txt
2.2 Conda Environments (Anaconda/Miniconda)
Conda acts as both a package manager and an environment manager. It can install Python itself and non-Python dependencies (C, CUDA, R). It uses a SAT solver for advanced dependency resolution, making it perfect for Data Science and AI/ML.
# Creating a Conda environment [14]
print("# Create new environment with Python 3.11")
print("conda create --name ai-lab python=3.11")

# Activating and Deactivating [14]
print("# Activate")
print("conda activate ai-lab")
print("# Deactivate")
print("conda deactivate")

# Installing packages (can install Python and system libs) [14, 15]
print("conda install numpy pandas matplotlib")

# Exporting and Recreating the environment [14]
print("# Export to YAML file")
print("conda env export > environment.yml")
print("# Recreate environment")
print("conda env create -f environment.yml")
2.3 Co-existence Demo
Both venv and Conda environments can exist without conflict on the same machine because each is an isolated folder containing its own dependencies and interpreter.
# Demo showing separate Django versions co-existing [18, 19]
print("# Venv (Web Project): Django 3.2 installed")
# python -m venv webenv
# webenv\Scripts\activate
# pip install django==3.2
# python -m django --version # Output: 3.2

print("# Conda (ML Project): Django 4.0 installed")
# conda activate ml-env
# conda install django=4.0
# python -m django --version # Output: 4.0

--------------------------------------------------------------------------------
Section 3: Variables, Data Types & Memory Management
3.1 Basic Data Types and Type Casting
Python is dynamically typed. Basic types include int, float, str, and bool. Type casting uses built-in functions like int(), float(), str(), and bool().
# Type Casting Examples [22]
age_str = "25"
age_num = int(age_str)
print(age_num + 5)      # 30

pi = 3.99
print(int(pi))          # 3 (truncates decimals)

# Boolean casting: 0, empty strings (""), [], and None are Falsy [22, 23]
print(bool(0))    # False
print(bool("Hi")) # True
3.2 Type Hints (Function Annotations)
Type hints improve readability and enable static analysis by tools like mypy. They are not enforced at runtime.
from typing import List, Dict, Optional, Union, Any, Callable, Tuple, Set
# For Python 3.9+, standard list[int] syntax is preferred [28]

def add_numbers(a: int, b: int) -> int:
    return a + b

def user_profile() -> Dict[str, Union[str, int]]:
    # Union allows multiple types [29]
    return {"name": "Aravind", "age": 25}

def get_discount(price: float, coupon: Optional[str]) -> float:
    # Optional[T] is shorthand for Union[T, None] [29]
    return price * 0.9 if coupon == "NEW10" else price

def apply_operation(x: int, y: int, func: Callable[[int, int], int]) -> int:
    # Callable hint: input types (int, int), return type (int) [29]
    return func(x, y)

point: Tuple[int, int] = (10, 20)
3.3 Memory Management: Reference Counting & GC
Python primarily uses reference counting; memory is freed when the reference count drops to zero. The Garbage Collector (GC) handles cyclic references (objects pointing to each other).
import sys, gc

# Example 1: Reference Counting [30]
x = [1, 3, 6]
y = x
# sys.getrefcount(x) increases as more names point to it.

del y
# Ref count decreases. When it hits zero, the object is destroyed.

# Example 2: Cyclic References and GC [32, 33]
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

n1 = Node("A")
n2 = Node("B")
n1.next = n2
n2.next = n1  # Cycle created
del n1, n2 # Objects are not freed because ref count > 0 due to the cycle [33]
unreachable = gc.collect()
print(f"Unreachable objects collected: {unreachable}")

# del statement removes the name/reference, not the object itself [34]
a = 100
b = a
del a # Name 'a' is removed
print(f"b still exists: {b}")
3.4 Context Managers and Weak References
Context Managers (via with statement) guarantee safe resource management (setup via __enter__, cleanup via __exit__), regardless of exceptions.
Weak References (weakref) allow referencing an object without increasing its reference count, ensuring the object can still be garbage collected.
import weakref
from contextlib import contextmanager

# 1. Standard Context Manager (File) [37]
with open("temp_data.txt", "w") as f:
    f.write("Test log") # File is automatically closed by __exit__() [37]

# 2. Custom Function-based Context Manager (using yield) [39, 41]
@contextmanager
def managed_resource():
    print("Setup (Before yield)")
    yield "Resource Handle" # Control passes to 'with' block [41]
    print("Teardown (After yield)")

with managed_resource() as r:
    print("Using resource:", r)

# 3. Weak Reference Example [40]
class Data: pass
obj = Data()
r = weakref.ref(obj) # Weak reference does not increase refcount
del obj
print(f"Object accessible after del: {r()}") # None (object destroyed)
3.5 Shallow vs Deep Copying
Operation
Outer Object
Inner Objects
Source
Assignment =
Shared
Shared
Shallow Copy
New
Shared
Deep Copy
New
New
Shallow copy (copy.copy()) creates a new container but shares references to inner, nested objects. Deep copy (copy.deepcopy()) recursively duplicates all nested objects, ensuring full independence.
import copy

# Nested structure (list containing mutable lists) [44]
a = [[1, 3], [6, 46]]

# 1. Shallow Copy (Inner list SHARED) [44]
b_shallow = copy.copy(a)
b_shallow = 99
print(f"A (Shallow Copy modification): {a}") # [[3, 34], [6, 46]]

# Reset A
a = [[1, 3], [6, 46]]

# 2. Deep Copy (Inner list COPIED) [45]
b_deep = copy.deepcopy(a)
b_deep = 999
print(f"A (Deep Copy modification): {a}") # [[1, 3], [6, 46]]

--------------------------------------------------------------------------------
Section 4: Operators and Control Flow
4.1 Arithmetic and Comparison Operators
Python differences from Java/C++: True division (/) always returns a float, and Floor division (//) provides the integer quotient. Exponentiation uses **. Comparisons can be chained (e.g., 5<x<10).
# Arithmetic Operators [47]
print(f"True Division: 5 / 2 = {5 / 2}") # 2.5
print(f"Floor Division: 5 // 2 = {5 // 2}") # 2
print(f"Exponentiation: 2 ** 3 = {2 ** 3}") # 8

# Chained Comparisons [49]
x = 7
print(f"5 < x < 10: {5 < x < 10}")
4.2 Logical and Identity Operators
Python uses keywords (and, or, not). Identity (is) checks if two variables refer to the exact same object in memory, unlike == which checks value equality.
# Logical Operators [51]
x = 5
print(f"x > 0 and x < 10: {x > 0 and x < 10}")

# Identity vs Equality [52]
a = [1, 3]
b = [1, 3]
c = a
print(f"a == b (Value Equality): {a == b}") # True
print(f"a is b (Identity Check): {a is b}") # False (different objects)
print(f"a is c: {a is c}") # True (same object)

# Membership Operator ('in') [53]
nums = [1, 3]
print(f"2 in nums: {2 in nums}")
4.3 The Walrus Operator (:=)
The assignment expression (:=) allows assigning a value to a variable within an expression.
# Walrus in a conditional statement [54]
if (n := len("Python")) > 5:
    print(f"String is too long, length: {n}")

# Walrus in a list comprehension [55]
nums = [56-60]
result = [half for n in nums if (half := n // 2) > 10]
# Walkthrough: 10/2=5(X), 25/2=12(V), 30/2=15(V), 5/2=2(X), 60/2=30(V) [61]
print(f"Walrus comprehension result: {result}")
# Output: [58, 62, 63] [64]
4.4 Control Flow: if/elif/else and Loops
Python uses indentation to define code blocks, and elif instead of else if. for loops iterate over sequences (like Java's enhanced for-each).
# if/elif/else syntax [65]
x = 10
if x > 0:
    print("Positive")
elif x == 0:
    print("Zero")
else:
    print("Negative")

# Iterating using range() [66]
print("Range loop (2, 10, step 2):")
for i in range(2, 10, 2):  # 2, 4, 6, 8 (stop index 10 is exclusive) [66]
    print(i)
4.5 Unique Python Feature: else with Loops
The else clause attached to a for or while loop executes only if the loop finishes normally (i.e., not terminated by a break statement). This is often used for "not found" checks.
# Example 1: Loop finishes normally (else runs) [69]
for i in range(3):
    print(i)
else:
    print("Loop finished without break") # This runs

# Example 2: Loop exits via break (else is skipped) [69]
for i in range(5):
    if i == 3:
        break
    print(i)
else:
    print("Loop finished without break") # This is skipped

--------------------------------------------------------------------------------
Section 5: Comprehensions and Generators
5.1 List Comprehensions
List comprehensions provide a concise way to create lists: [expression for item in iterable if condition].
nums = [1, 3, 6, 46, 59]

# Example 1: Squares of numbers [71]
squares = [n*n for n in nums]
print(f"Squares: {squares}")

# Example 2: Filtering evens [71]
evens = [n for n in nums if n % 2 == 0]
print(f"Evens: {evens}")

# Example 3: Nested Loops (Flattening) [72, 73]
matrix = [[1, 3, 6], [46, 59, 74]]
flat = [num for row in matrix for num in row]
print(f"Flattened list: {flat}")
5.2 Dictionary and Set Comprehensions
Dictionary: {key_expr: value_expr for item in iterable}. Set: {expr for item in iterable}.
# Dictionary Comprehension: Key-Value Pairs [75]
nums = [1, 3, 6]
squares_dict = {n: n*n for n in nums}
print(f"Dict Squares: {squares_dict}")

# Set Comprehension: Unique items [77]
nums_dupes = [1, 3, 3, 6, 6, 46]
unique_squares = {n*n for n in nums_dupes}
print(f"Set Squares (unordered): {unique_squares}")
5.3 Generator Expressions and Memory Efficiency
Generator expressions use () instead of []. They use lazy evaluation, producing items one at a time, making them highly memory-efficient for large datasets or streams.
import sys

# List Comprehension (Eager - stores all items) [80]
list_squares = [n*n for n in range(5)]

# Generator Expression (Lazy - computes on demand) [78, 80]
gen_squares = (n*n for n in range(5))

print(f"List size (approx): {sys.getsizeof(list_squares)} bytes")
print(f"Generator size (tiny recipe): {sys.getsizeof(gen_squares)} bytes")

# Generator pipeline for memory efficiency (only one item in memory at a time) [81]
nums = range(1_000_000)
# Sum of cubes of numbers divisible by 7 (lazy computation)
total = sum(n**3 for n in nums if n % 7 == 0)

--------------------------------------------------------------------------------
Section 6: Data Structures: List, Tuple, Set, Dictionary
6.1 Lists (Mutable Sequences)
Lists are ordered, mutable collections. Operations like inserting/deleting at the front (insert(0, x), pop(0)) are O(n) due to shifting elements.
# Slicing and Negative Indexing [85]
fruits = ["apple", "banana", "cherry", "date"]
print(f"Last element: {fruits[-1]}")
print(f"Reversed list: {fruits[::-1]}")

# Adding and Removing [86, 87]
lst = [1, 3]
lst.append(3)           # [1, 3, 6]
lst.insert(1, 5)        # [1, 3, 6, 59]
lst.extend([8, 74])      # [1, 3, 6, 8, 59, 74]
lst.remove(5)           # Removes first 5 [87]
lst.pop()               # Removes last (7) [88]
print(f"List after ops: {lst}")

# Sorting: sort() mutates in place; sorted() returns a new list [89]
nums = [1, 6, 46]
nums.sort()
print(f"Sorted in place: {nums}")
6.2 Tuples (Immutable Sequences)
Tuples are ordered, immutable sequences. They are faster and use less memory than lists, and are hashable (can be dictionary keys).
# Creating Tuples (Note: comma required for single element: (42,) ) [92, 93]
t1 = (1, 2, 3)
t_packed = 4, 5, 6 # Tuple packing [92]

# Immutability check
# t1 = 10 # Raises TypeError [94]

# Tuple Unpacking [94]
x, y, z = t1
print(f"Unpacked: {x}, {y}, {z}")

# Extended Unpacking (using *) [91]
data = (1, 2, 3, 4, 5)
a, *b, c = data
print(f"Extended Unpacking: a={a}, b={b} (list), c={c}")

# Swapping variables without temp [95]
a, b = 10, 20
a, b = b, a
print(f"Swapped: a={a}, b={b}")
6.3 Sets and Frozen Sets
Sets are unordered collections of unique elements, supporting fast membership checks (O(1)). Sets are mutable. Frozen Sets are immutable and hashable, usable as dictionary keys.
# Creating Sets [96]
fruits = {"apple", "banana", "cherry", "apple"}
print(f"Set (duplicates removed): {fruits}")

# Set Operations (Mathematical) [100, 101]
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(f"Union (a | b): {a | b}")         # All unique elements [100]
print(f"Intersection (a & b): {a & b}") # Common elements [100]
print(f"Difference (a - b): {a - b}")   # In a but not in b [102]
print(f"Subset (a <= b): {a <= b}")     # Check if a is contained in b [103]

# Frozen Set Example [104]
fs = frozenset([1, 3])
# fs.add(3) # Raises AttributeError [99]
d = {fs: "data"} # Can be used as a key [99]
6.4 Dictionaries (Key-Value Pairs)
Dictionaries are mutable collections of key-value pairs. Keys must be unique and immutable. Dictionaries maintain insertion order (Python 3.7+).
# Creating and Accessing [108]
student = {"name": "Alice", "age": 22, "major": "CS"}
print(f"Access (direct): {student['name']}")
print(f"Access (safe): {student.get('grade', 'N/A')}") # Use .get() to avoid KeyError [109]

# Updating and Adding [109]
student["age"] = 23 # Update
student["email"] = "alice@example.com" # Add new
student.update({"major": "Physics"}) # Merge/overwrite [109]

# Deletion [110]
age = student.pop("age") # Remove and return value (safer than del)
del student["major"] # Delete specific key
print(f"Student keys left: {student.keys()}")

# Dictionary Comprehension [111]
nums = [1, 3, 6]
squares_comp = {n: n**2 for n in nums}
print(f"Dict Comp: {squares_comp}")
6.5 Advanced Dictionary Tools (collections)
These tools provide enhanced functionality over standard dicts.
from collections import defaultdict, Counter, ChainMap, namedtuple
from types import MappingProxyType
from dataclasses import dataclass, field

# 1. defaultdict: Provides default value for missing keys [113, 114]
count = defaultdict(int)
words = ["apple", "banana", "apple"]
for w in words: count[w] += 1
print(f"defaultdict: {count}")

# 2. Counter: Subclass for counting hashable objects [115, 116]
freq = Counter(words)
print(f"Counter (Most common): {freq.most_common(1)}")

# 3. ChainMap: Combines multiple dictionaries (searches layers in order) [117, 118]
defaults = {"theme": "light"}
user_settings = {"theme": "dark"}
config = ChainMap(user_settings, defaults)
print(f"ChainMap theme: {config['theme']}")

# 4. Named Tuples: Immutable record types [119, 120]
Point = namedtuple('Point', ['x', 'y'])
p = Point(10, 20)
print(f"Named Tuple: {p.x}")

# 5. Dataclasses: Structured data with auto-generated methods [121, 122]
@dataclass
class Person:
    name: str
    age: int = 18
    skills: list = field(default_factory=list) # Use default_factory for mutable defaults [123]

--------------------------------------------------------------------------------
Section 7: Functions and Advanced Concepts
7.1 Function Definition and Arguments
Functions are declared with def. They are first-class citizens (can be assigned, passed, returned). Python supports multiple returns (packed as a tuple).
# Multiple Returns (Tuple Packing/Unpacking) [130]
def min_max(numbers):
    return min(numbers), max(numbers) # Returns a tuple
low, high = min_max([1, 3, 9])
print(f"Min: {low}, Max: {high}")

# Default Parameters and Keyword Arguments [127, 130]
def describe_pet(name, species="Dog"): # Default
    print(f"{name} is a {species}")

describe_pet(species="Cat", name="Whiskers") # Keyword Arguments (order irrelevant) [131]
describe_pet("Buddy") # Uses default
7.2 Variable-Length Arguments (*args, **kwargs)
*args collects extra positional arguments into a tuple. **kwargs collects extra keyword arguments into a dictionary.
# *args: collects positional arguments into a tuple [132]
def add_all(*numbers):
    return sum(numbers)
print(f"Sum: {add_all(1, 2, 3, 4)}")

# **kwargs: collects keyword arguments into a dict [132]
def show_profile(name, **info):
    print(f"Name: {name}. Other info: {info}")
show_profile(name="Aravind", age=25, role="Dev")
7.3 The Mutable Default Parameter Trap ⚠️
Mutable objects (lists, dicts, sets) should not be used as default parameters, as they are shared across all function calls.
# ❌ INCORRECT: Mutable default [] is shared [135]
# def add_item_bad(item, basket=[]):
#     basket.append(item)
# print(add_item_bad("apple")) # ['apple']
# print(add_item_bad("banana")) # ['apple', 'banana'] <-- shared list! [135]

# ✅ CORRECT: Use None and initialize inside the function [136]
def add_item_safe(item, basket=None):
    if basket is None:
        basket = [] # New list created on each call
    basket.append(item)
    return basket

print(f"Safe call 1: {add_item_safe('apple')}")
print(f"Safe call 2: {add_item_safe('banana')}")
7.4 Lambda Functions and Functional Programming
Lambda functions are small, anonymous, one-line functions (lambda args: expression). They are often used with Higher-Order Functions (HOFs) like map, filter, and sorted.
from functools import reduce

# Lambda syntax: lambda arguments: expression [137]
square = lambda x: x**2
print(f"Lambda square(5): {square(5)}")

nums = [1, 3, 6, 46]
# map: applies lambda to all elements [138]
squared = list(map(lambda x: x**2, nums))

# filter: keeps elements returning True [138]
evens = list(filter(lambda x: x % 2 == 0, nums))

# sorted: uses lambda as a key function to sort by custom criteria [140]
points = [(1, 5), (2, 3), (4, 1)]
sorted_points = sorted(points, key=lambda p: p[1]) # Sort by y-value
print(f"Sorted by key: {sorted_points}")
7.5 Higher-Order Functions (HOFs) and Closures
A HOF takes or returns a function. Closures are inner functions that remember variables from their outer (enclosing) scope even after the outer function finishes executing. This enables state maintenance and decorators.
# HOF: Function that returns a function (Closure Example) [146]
def make_multiplier(n): # n is captured by the closure
    def multiplier(x):
        return x * n # Inner function remembers 'n'
    return multiplier

double = make_multiplier(2)
print(f"Double 10: {double(10)}") # 20

# Closure for maintaining state (Counter Example) [147]
def counter():
    count = 0
    def increment():
        nonlocal count # Needed to modify enclosing variable [147]
        count += 1
        return count
    return increment

c = counter()
print(f"Counter: {c()}")
print(f"Counter: {c()}") # State persists [148]
7.6 Decorators and Decorator Factories
Decorators are HOFs that wrap functions to add functionality using the @ syntax. Decorators with arguments (like @repeat(3)) are called Decorator Factories.
import time
from functools import wraps

# Example 1: Basic Decorator (Wrapping behavior) [152]
def greet_decorator(func):
    @wraps(func) # Preserves name/docstring [153]
    def wrapper(*args, **kwargs):
        print("Before the function runs")
        result = func(*args, **kwargs)
        print("After the function runs")
        return result
    return wrapper

@greet_decorator
def say_hello():
    print("Hello!")
say_hello()

# Example 2: Decorator Factory (Decorator with Arguments) [154]
def repeat(n):
    def decorator(func): # Closure remembers 'n' [154]
        @wraps(func)
        def wrapper(*args, **kwargs): # Handles function arguments
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator

@repeat(3)
def greet():
    print("Hi")
greet() # Runs 'Hi' three times [151]

--------------------------------------------------------------------------------
Section 8: Modules, Packages, and pip
8.1 Modules and Importing
A module is a single .py file. A package is a directory containing modules, marked by an __init__.py file.
import math
from datetime import datetime, timedelta

print(f"Square root of 16: {math.sqrt(16)}") # Basic import [157]

now = datetime.now()
future = now + timedelta(days=7)
print(f"One week later: {future}") # Standard Library usage [159]
8.2 Writing Custom Modules and Packages
The __init__.py file is essential for marking a folder as a package and is often used to expose a clean public API via re-exporting functions (e.g., from .algebra import add).
# Example: Using the __name__ guard [161]
# Runs only if the file is executed directly (python file.py) [161]
# if __name__ == "__main__":
#     print("Testing module...")

# Example: Re-exporting in __init__.py [160]
# In mymath/__init__.py:
# from .algebra import add
# from .geometry import area_circle
# Usage: from mymath import add, area_circle
8.3 Package Management with pip
pip is Python's package installer, fetching libraries from PyPI. Virtual environments (venv or Conda) are essential for isolating dependencies.
# Installing and managing packages [162]
print("pip install requests")
print("pip install requests==2.31.0")

# Reproducibility (locking dependency versions) [165]
print("# Freeze installed packages to a file")
print("pip freeze > requirements.txt")
print("# Install from the file")
print("pip install -r requirements.txt")

--------------------------------------------------------------------------------
Section 9: Object-Oriented Programming (OOP)
9.1 Classes, Attributes, and the Constructor
The constructor is defined by __init__(self, ...). self refers to the current instance (like this in Java). Class variables are shared across instances.
class Dog:
    species = "Canine" # Class variable (shared) [169]

    def __init__(self, name, breed):
        self.name = name   # Instance attribute [166]
        self.breed = breed

    def bark(self):
        return f"{self.name} says Woof!"

dog1 = Dog("Buddy", "Labrador")

# Modification via Class Name affects all instances [170]
Dog.species = "Domestic Canine"
print(f"{dog1.name} is a {dog1.species}")
9.2 Inheritance, Polymorphism, and Encapsulation
Python supports both single and multiple inheritance. Encapsulation uses naming conventions: Protected (_attribute) and Private (__attribute, which is name-mangled). Abstraction uses Abstract Base Classes (ABC) and @abstractmethod to enforce interfaces.
from abc import ABC, abstractmethod

# Abstract Base Class (Abstraction) [176, 177]
class MLModel(ABC):
    def __init__(self, name):
        self._trained = False # Protected attribute (convention) [172]

    @abstractmethod # Forces subclasses to implement [175]
    def train(self, data):
        pass

# Utility Class (for Multiple Inheritance) [177-179]
class Logger:
    def log(self, message):
        print(f"[LOG]: {message}")

class LinearRegressionModel(MLModel, Logger): # Multiple Inheritance [171, 180]
    def __init__(self, name, learning_rate):
        super().__init__(name) # Calls parent (MLModel) constructor [181, 182]
        self.__weights = None # Private attribute (name-mangled) [173]

    def train(self, data):
        self._trained = True
        self.log(f"{self.name} trained.") # Use inherited method

# Polymorphic Function [183]
def evaluate_model(model: MLModel, data, input_data):
    # Works regardless of subclass type
    model.train(data)
    # Polymorphism: calls the appropriate train() method
9.3 Advanced OOP Features
a) Magic Methods (Dunder Methods)
Methods like __len__ and __repr__ allow classes to emulate built-in types.
class Dataset:
    def __init__(self, name, samples):
        self._samples = samples

    def __len__(self): # Defines behavior for len(obj) [185]
        return len(self._samples)

    def __repr__(self): # Unambiguous string representation for debugging [185]
        return f"Dataset(name='{self.name}', samples={len(self)})"

ds = Dataset("MyData", ["a", "b"])
print(f"Length: {len(ds)}") # Calls __len__
b) Class Method vs Static Method
Decorator
Argument
Purpose
Source
@classmethod
cls
Alternate constructors, uses class state
@staticmethod
None
Utility functions, no access to instance or class
class DataFactory:
    @staticmethod
    def validate_sample(sample): # Utility function [188]
        return isinstance(sample, str)

    @classmethod
    def from_config(cls, config_dict): # Alternate constructor [187]
        return cls(config_dict["name"], config_dict["data"])
c) Property Decorators (@property, @setter)
Provide controlled access (getters and setters) to internal attributes, making methods behave like attributes. You can use @property alone for read-only access.
class Circle:
    def __init__(self, radius):
        self._radius = radius # Internal storage

    @property
    def radius(self): # Acts as the getter [189]
        return self._radius

    @radius.setter
    def radius(self, value): # Controlled setter (validation logic) [190]
        if value < 0:
            raise ValueError("Radius cannot be negative")
        self._radius = value

c = Circle(5)
print(f"Radius (property access): {c.radius}")
c.radius = 10 # Uses the setter method
````
