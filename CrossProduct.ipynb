```python
# Cross product function (manual implementation)
def cross_product(a, b):
    """
    Compute cross product of two 3D vectors a and b.
    a, b: lists or tuples of length 3
    Returns: list of length 3
    """
    return [
        a[1]*b[2] - a[2]*b[1],  # x-component
        a[2]*b[0] - a[0]*b[2],  # y-component
        a[0]*b[1] - a[1]*b[0]   # z-component
    ]

# Example: Dr. A (urban cardiologist, critical patients)
a = [5, 8, 9]
# Example: Dr. B (rural pediatrician, moderate patients)
b = [3, 2, 5]

c = cross_product(a, b)
print("Cross Product:", c)
```

```python
import numpy as np

# Example provider feature vectors (3D)
# Let's assume each dimension represents: [specialization, location, patient_profile]
provider1 = np.array([0.8, 0.3, 0.5])
provider2 = np.array([0.4, 0.9, 0.2])

# 1️⃣ Cross product
cross_prod = np.cross(provider1, provider2)
print("Cross Product Vector:", cross_prod)

# 2️⃣ Magnitude of the cross product
magnitude = np.linalg.norm(cross_prod)
print("Magnitude of Cross Product:", magnitude)

# 3️⃣ Angle between the two vectors (in degrees)
# Formula: angle = arccos((A.B) / (||A|| * ||B||))
dot_product = np.dot(provider1, provider2)
norm1 = np.linalg.norm(provider1)
norm2 = np.linalg.norm(provider2)
cos_theta = dot_product / (norm1 * norm2)
angle_rad = np.arccos(np.clip(cos_theta, -1.0, 1.0))  # clip to avoid numerical errors
angle_deg = np.degrees(angle_rad)
print("Angle between vectors (degrees):", angle_deg)
```

```python
import math

# Example provider vectors (3D)
provider1 = [0.8, 0.3, 0.5]
provider2 = [0.4, 0.9, 0.2]

# 1️⃣ Cross product (manual calculation)
def cross_product(a, b):
    return [
        a[1]*b[2] - a[2]*b[1],
        a[2]*b[0] - a[0]*b[2],
        a[0]*b[1] - a[1]*b[0]
    ]

cross_prod = cross_product(provider1, provider2)
print("Cross Product Vector:", cross_prod)

# 2️⃣ Magnitude of a vector
def magnitude(v):
    return math.sqrt(sum([x**2 for x in v]))

cross_magnitude = magnitude(cross_prod)
print("Magnitude of Cross Product:", cross_magnitude)

# 3️⃣ Angle between two vectors (in degrees)
def dot_product(a, b):
    return sum([a[i]*b[i] for i in range(len(a))])

dot = dot_product(provider1, provider2)
mag1 = magnitude(provider1)
mag2 = magnitude(provider2)

cos_theta = dot / (mag1 * mag2)
# Clip cos_theta between -1 and 1 to avoid math domain errors due to floating point precision
cos_theta = max(min(cos_theta, 1), -1)

angle_rad = math.acos(cos_theta)
angle_deg = math.degrees(angle_rad)
print("Angle between vectors (degrees):", angle_deg)
```

```python
import math
import matplotlib.pyplot as plt

# -----------------------------
# 1️⃣ Define provider vectors
# -----------------------------
# Features: [Specialization, Location, Patient Profile]
providers = {
    "P1": [1.0, 0.0, 0.5],
    "P2": [0.0, 1.0, 0.5],
    "P3": [1.0, 1.0, 0.0],
    "P4": [0.5, 0.5, 1.0],
    "P5": [0.2, 0.8, 0.3]
}

# -----------------------------
# 2️⃣ Functions for vector math
# -----------------------------
def cross_product(a, b):
    return [
        a[1]*b[2] - a[2]*b[1],
        a[2]*b[0] - a[0]*b[2],
        a[0]*b[1] - a[1]*b[0]
    ]

def dot_product(a, b):
    return sum([a[i]*b[i] for i in range(len(a))])

def magnitude(v):
    return math.sqrt(sum([x**2 for x in v]))

def angle_between(a, b):
    dot = dot_product(a, b)
    mag_a = magnitude(a)
    mag_b = magnitude(b)
    cos_theta = dot / (mag_a * mag_b)
    cos_theta = max(min(cos_theta, 1), -1)  # clip for numerical safety
    return math.degrees(math.acos(cos_theta))

# -----------------------------
# 3️⃣ Compute pairwise relationships
# -----------------------------
print("Pairwise Cross Products, Magnitudes, and Angles:")
provider_names = list(providers.keys())
for i in range(len(provider_names)):
    for j in range(i+1, len(provider_names)):
        p1 = providers[provider_names[i]]
        p2 = providers[provider_names[j]]
        cp = cross_product(p1, p2)
        mag = magnitude(cp)
        angle = angle_between(p1, p2)
        print(f"{provider_names[i]} x {provider_names[j]} -> Cross: {cp}, |Cross|: {mag:.3f}, Angle: {angle:.2f}°")

# -----------------------------
# 4️⃣ Visualize in 3D
# -----------------------------
fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')

for name, vec in providers.items():
    ax.quiver(0, 0, 0, vec[0], vec[1], vec[2], length=1.0, normalize=True)
    ax.text(vec[0], vec[1], vec[2], name, size=12, zorder=1)

ax.set_xlabel('Specialization')
ax.set_ylabel('Location')
ax.set_zlabel('Patient Profile')
ax.set_title('3D Provider Vector Embedding')
plt.show()
```

```python
import math

# -----------------------------
# 1️⃣ Provider vectors (3D)
# -----------------------------
providers = {
    "P1": [1.0, 0.0, 0.5],
    "P2": [0.0, 1.0, 0.5],
    "P3": [1.0, 1.0, 0.0],
    "P4": [0.5, 0.5, 1.0],
    "P5": [0.2, 0.8, 0.3]
}

# -----------------------------
# 2️⃣ Vector math functions
# -----------------------------
def cross_product(a, b):
    return [
        a[1]*b[2] - a[2]*b[1],
        a[2]*b[0] - a[0]*b[2],
        a[0]*b[1] - a[1]*b[0]
    ]

def dot_product(a, b):
    return sum([a[i]*b[i] for i in range(len(a))])

def magnitude(v):
    return math.sqrt(sum([x**2 for x in v]))

def angle_between(a, b):
    dot = dot_product(a, b)
    mag_a = magnitude(a)
    mag_b = magnitude(b)
    cos_theta = dot / (mag_a * mag_b)
    cos_theta = max(min(cos_theta, 1), -1)
    return math.degrees(math.acos(cos_theta))

# -----------------------------
# 3️⃣ Complementary provider suggestion
# -----------------------------
def suggest_complementary(provider_name, top_n=2):
    if provider_name not in providers:
        print("Provider not found.")
        return []
    
    base_vec = providers[provider_name]
    scores = []
    
    for name, vec in providers.items():
        if name == provider_name:
            continue
        cp_mag = magnitude(cross_product(base_vec, vec))
        angle = angle_between(base_vec, vec)
        # Combined score (simple: magnitude + normalized angle)
        score = cp_mag + angle/180  # normalizing angle roughly
        scores.append((name, score, cp_mag, angle))
    
    # Sort by score descending → most complementary first
    scores.sort(key=lambda x: x[1], reverse=True)
    
    print(f"Top {top_n} complementary providers for {provider_name}:")
    for s in scores[:top_n]:
        print(f"{s[0]} -> Score: {s[1]:.3f}, |Cross|: {s[2]:.3f}, Angle: {s[3]:.2f}°")
    
    return scores[:top_n]

# -----------------------------
# 4️⃣ Test the function
# -----------------------------
suggest_complementary("P1", top_n=3)
```
