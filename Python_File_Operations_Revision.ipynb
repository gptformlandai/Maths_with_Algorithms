
# 1. Basic File Operations
This section covers:
- Reading and writing text files
- Using `with open()` context manager
- Working with CSV files (reader, writer, DictReader, DictWriter, quoting, delimiter, escapechar)


```

# Writing and reading a text file
with open('example.txt', 'w') as f:
    f.write("Hello, Aravind!
Welcome to Python file operations.")

with open('example.txt', 'r') as f:
    content = f.read()

print(content)
```

```

# CSV reader and writer with optional parameters
import csv

rows = [['Name', 'Age'], ['Aravind', 28], ['Meera', 32]]
with open('people.csv', 'w', newline='') as f:
    writer = csv.writer(f, delimiter='|', quotechar='"', quoting=csv.QUOTE_ALL)
    writer.writerows(rows)

with open('people.csv', 'r') as f:
    reader = csv.reader(f, delimiter='|', quotechar='"')
    for row in reader:
        print(row)
```

```

# DictWriter and DictReader
with open('people_dict.csv', 'w', newline='') as f:
    writer = csv.DictWriter(f, fieldnames=['Name', 'Age'])
    writer.writeheader()
    writer.writerow({'Name': 'Aravind', 'Age': 28})

with open('people_dict.csv', 'r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row)
```


# 2. Advanced File Operations
This section covers:
- JSON handling (load, dump, loads, dumps)
- Working with different encodings (UTF-8, UTF-16)
- Reading/writing binary files (basic and with struct)


```

# JSON handling
import json

config = {"model": "RandomForest", "n_estimators": 100}
with open('config.json', 'w') as f:
    json.dump(config, f, indent=4)

with open('config.json', 'r') as f:
    loaded = json.load(f)

print(loaded)
```

```

# Encoding example
config = {"language": "日本語", "greeting": "こんにちは"}
with open('utf8_config.json', 'w', encoding='utf-8') as f:
    json.dump(config, f, ensure_ascii=False)

with open('utf8_config.json', 'r', encoding='utf-8') as f:
    print(json.load(f))

with open('utf16_config.json', 'w', encoding='utf-16') as f:
    json.dump(config, f, ensure_ascii=False)
```

```

# Binary file operations
binary_data = bytes([120, 3, 255, 0, 100])
with open('sample_binary.bin', 'wb') as f:
    f.write(binary_data)

with open('sample_binary.bin', 'rb') as f:
    data = f.read()
    print(list(data))
```

```

# Structured binary data with struct
import struct
packed = struct.pack('iiff', 42, 7, 3.14, 2.718)
with open('structured.bin', 'wb') as f:
    f.write(packed)

with open('structured.bin', 'rb') as f:
    unpacked = struct.unpack('iiff', f.read())
    print(unpacked)
```


# 3. Path Manipulation with pathlib
This section covers:
- Creating and checking paths
- Reading/writing files
- Listing directory contents


```

from pathlib import Path

# Create a path and write to a file
p = Path('notes')
p.mkdir(exist_ok=True)
file_path = p / 'note.txt'
file_path.write_text("This is a note using pathlib.")

# Read the file
print(file_path.read_text())

# List contents of the directory
for item in p.iterdir():
    print(item)
```
