```python
import numpy as np

# Dot product manually for 2D
def dot2D(a, b):
    return a[0]*b[0] + a[1]*b[1]

# Dot product manually for 3D
def dot3D(a, b):
    return a[0]*b[0] + a[1]*b[1] + a[2]*b[2]

# Examples
v2a = [1, 2]
v2b = [3, 4]
print("Dot product (2D):", dot2D(v2a, v2b))

v3a = [1, 2, 3]
v3b = [4, 5, 6]
print("Dot product (3D):", dot3D(v3a, v3b))
```

```python
import math

def cosine_similarity(a, b):
    # dot product
    dot = sum(x*y for x, y in zip(a, b))
    # magnitudes
    mag_a = math.sqrt(sum(x*x for x in a))
    mag_b = math.sqrt(sum(x*x for x in b))
    return dot / (mag_a * mag_b)

# Example
v1 = [3, 4]
v2 = [4, 3]
cos_sim = cosine_similarity(v1, v2)
angle = math.degrees(math.acos(cos_sim))

print("Cosine similarity:", cos_sim)
print("Angle (degrees):", angle)
```

```python
import matplotlib.pyplot as plt

def plot_vectors(v1, v2):
    plt.figure(figsize=(6,6))
    ax = plt.gca()
    ax.set_aspect('equal')

    # Draw vectors
    plt.quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1, color='r', label=f"v1 {v1}")
    plt.quiver(0, 0, v2[0], v2[1], angles='xy', scale_units='xy', scale=1, color='b', label=f"v2 {v2}")

    # Axis settings
    plt.xlim(-1, max(v1[0], v2[0])+2)
    plt.ylim(-1, max(v1[1], v2[1])+2)
    plt.grid(True)
    plt.legend()

    # Cosine similarity & angle
    cos_sim = cosine_similarity(v1, v2)
    angle = math.degrees(math.acos(cos_sim))
    plt.title(f"Cosine similarity = {cos_sim:.3f},  Angle = {angle:.2f}°")

    plt.show()

# Example
plot_vectors([3,4], [4,3])
```
