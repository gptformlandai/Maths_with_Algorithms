
### First-Class and Higher-Order Functions
Python treats functions as first-class citizens. You can assign them to variables, pass them as arguments, and return them from other functions.
        

```

def greet(name):
    return f"Hello, {name}!"

say_hello = greet
print(say_hello("Aravind"))

def shout(text):
    return text.upper()

def whisper(text):
    return text.lower()

def greet_style(func):
    return func("Hello, Aravind")

print(greet_style(shout))
print(greet_style(whisper))
        
```


### Lambda Functions
Lambda functions are anonymous functions defined with the `lambda` keyword. They are useful for short, throwaway functions.
        

```

square = lambda x: x * x
print(square(5))

nums = [1, 2, 3, 4]
squared = list(map(lambda x: x**2, nums))
print(squared)
        
```


### Classes and Objects
Classes are blueprints for creating objects. Objects are instances of classes.
        

```

class Dog:
    species = "Canine"  # Class variable

    def __init__(self, name):
        self.name = name  # Instance variable

    def bark(self):
        return f"{self.name} says Woof!"

dog1 = Dog("Buddy")
print(dog1.bark())
print(dog1.species)
        
```


### OOP Concepts: Inheritance, Polymorphism, Encapsulation, Abstraction
This section demonstrates inheritance, polymorphism, encapsulation, and abstraction using a model framework.
        

```

from abc import ABC, abstractmethod

class MLModel(ABC):
    def __init__(self, name):
        self.name = name
        self._trained = False

    @abstractmethod
    def train(self, data):
        pass

    @abstractmethod
    def predict(self, input_data):
        pass

class Logger:
    def log(self, message):
        print(f"[LOG]: {message}")

class LinearRegressionModel(MLModel, Logger):
    def __init__(self, name, learning_rate):
        super().__init__(name)
        self.learning_rate = learning_rate
        self.__weights = None

    def train(self, data):
        self.__weights = [0.5] * len(data[0])
        self._trained = True
        self.log(f"{self.name} trained with learning rate {self.learning_rate}")

    def predict(self, input_data):
        if not self._trained:
            raise Exception("Model not trained yet!")
        return sum([w * x for w, x in zip(self.__weights, input_data)])

class DecisionTreeModel(MLModel, Logger):
    def __init__(self, name, max_depth):
        super().__init__(name)
        self.max_depth = max_depth
        self.__tree = None

    def train(self, data):
        self.__tree = {"root": "feature_1"}
        self._trained = True
        self.log(f"{self.name} trained with max depth {self.max_depth}")

    def predict(self, input_data):
        if not self._trained:
            raise Exception("Model not trained yet!")
        return "Class_A"

def evaluate_model(model: MLModel, data, input_data):
    model.train(data)
    prediction = model.predict(input_data)
    print(f"Prediction from {model.name}: {prediction}")

data = [[1, 2], [3, 4]]
input_data = [5, 6]

lr_model = LinearRegressionModel("LinearReg", 0.01)
dt_model = DecisionTreeModel("DecisionTree", 5)

evaluate_model(lr_model, data, input_data)
evaluate_model(dt_model, data, input_data)
        
```


### Classmethod, Staticmethod, Property and Setter
These decorators provide different ways to define methods and manage attribute access.
        

```

class Circle:
    def __init__(self, radius):
        self._radius = radius

    @property
    def radius(self):
        return self._radius

    @radius.setter
    def radius(self, value):
        if value < 0:
            raise ValueError("Radius cannot be negative")
        self._radius = value

    @staticmethod
    def area_formula():
        return "Area = π * r^2"

    @classmethod
    def unit_circle(cls):
        return cls(1)

c = Circle.unit_circle()
print(c.radius)
c.radius = 5
print(c.radius)
print(Circle.area_formula())
        
```
